# Autonomous Data Preprocessing Pipeline Demo

This notebook demonstrates how to run the autonomous data preprocessing pipeline locally.

## Features Demonstrated:
- Dataset ingestion and validation
- Automated data cleaning and preprocessing
- Feature engineering and transformation
- Exploratory Data Analysis (EDA)
- Train/test splitting
- PTMA autonomy metrics calculation

## Requirements:
- Python 3.8+
- Dependencies from requirements.txt
- Sample dataset (titanic.csv)

```python
# Install dependencies (uncomment if needed)
# !pip install -r ../requirements.txt

# Import required modules
import sys
import os
import pandas as pd
import json

# Add parent directory to path
sys.path.append('..')

# Import pipeline components
from pipeline_orchestrator import PipelineOrchestrator
from llm_agent import create_llm_agent

print("✅ Imports successful!")
```

## Loading Sample Dataset

Let's load the Titanic dataset for demonstration.

```python
# Load sample dataset
dataset_path = "../docs/sample-input.csv"

if os.path.exists(dataset_path):
    df = pd.read_csv(dataset_path)
    print(f"✅ Dataset loaded successfully!")
    print(f"Shape: {df.shape}")
    print(f"Columns: {list(df.columns)}")
    print("\nFirst 5 rows:")
    display(df.head())
else:
    print(f"❌ Dataset not found at {dataset_path}")
    print("Please ensure the sample dataset exists.")
```

## Initializing Pipeline Orchestrator

Create the pipeline orchestrator with LLM agent support.

```python
# Initialize LLM agent (set to None to disable)
llm_agent = create_llm_agent(use_llm=True, provider="groq")  # Options: groq, openai, anthropic

# Create pipeline orchestrator
orchestrator = PipelineOrchestrator(target_col="Survived", llm_agent=llm_agent)

print("✅ Pipeline orchestrator initialized!")
print(f"Target column: Survived")
print(f"LLM agent: {'Enabled' if llm_agent else 'Disabled'}")
```

## Running Complete Pipeline

Execute the full autonomous preprocessing pipeline. This will take 30-60 seconds.

```python
import time

print("🚀 Starting autonomous preprocessing pipeline...")
print("This will take 30-60 seconds depending on dataset size and LLM usage.")

start_time = time.time()

try:
    # Run the complete pipeline
    result = orchestrator.run(input_file=dataset_path)
    
    execution_time = time.time() - start_time
    
    print(f"\n✅ Pipeline completed successfully in {execution_time:.2f} seconds!")
    
except Exception as e:
    execution_time = time.time() - start_time
    print(f"\n❌ Pipeline failed after {execution_time:.2f} seconds")
    print(f"Error: {str(e)}")
    result = None
```

## Examining Pipeline Results

Let's examine the PTMA autonomy metrics and pipeline outputs.

```python
if result:
    # Extract PTMA metrics
    ptma_metrics = result.get('autonomy_metrics', {})
    
    print("📊 PTMA Autonomy Metrics:")
    print(f"Tasks completed: {ptma_metrics.get('tasks', 0)}")
    print(f"LLM prompts used: {ptma_metrics.get('prompts', 0)}")
    print(f"Corrections needed: {ptma_metrics.get('corrections', 0)}")
    print(f"PTMA Score: {ptma_metrics.get('PTMA', 0):.3f}")
    
    # Interpret autonomy level
    ptma_score = ptma_metrics.get('PTMA', 0)
    if ptma_score > 0.8:
        autonomy_level = "Excellent - Highly Autonomous"
    elif ptma_score > 0.6:
        autonomy_level = "Good - Moderately Autonomous"
    elif ptma_score > 0.4:
        autonomy_level = "Fair - Some Human Intervention"
    else:
        autonomy_level = "Needs Improvement - High Human Intervention"
    
    print(f"\n🤖 Autonomy Level: {autonomy_level}")
    
else:
    print("❌ No pipeline results available. Please check the error above.")
```

## Viewing Processed Data

Let's examine the processed dataset output.

```python
# Check if processed data is available
if hasattr(orchestrator, 'context') and hasattr(orchestrator.context, 'processed_data'):
    processed_df = orchestrator.context.processed_data
    
    print("📋 Processed Dataset Preview:")
    print(f"Shape: {processed_df.shape}")
    print(f"Columns: {list(processed_df.columns)}")
    print("\nFirst 5 rows:")
    display(processed_df.head())
    
    # Save sample output
    sample_output_path = "../docs/example-output.csv"
    processed_df.head(10).to_csv(sample_output_path, index=False)
    print(f"\n💾 Sample output saved to: {sample_output_path}")
    
else:
    print("❌ Processed data not available in orchestrator context.")
    print("This might indicate an issue with the pipeline execution.")
```
